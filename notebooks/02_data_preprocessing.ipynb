# 02 - Data Preprocessing & Validation

### Overview
- Parse dates into standard datetime representations.
- Ensure numeric data types for sales and store identifiers.
- Validate data consistency and absence of anomalies.
- Export cleaned dataset to `data/processed/train_processed2.csv`.


```python
import pandas as pd 
import numpy as np 
import matplotlib.pyplot as plt
import seaborn as sns 
```

```python
import os
base_dir = ".." if os.path.exists("../data") else "."
df = pd.read_csv(f"{base_dir}/data/processed/train_processed.csv")
df.head()
```

```python
df['date']=pd.to_datetime(df['date'])
```

```python
df.info()
```

```python
df['sales']=pd.to_numeric(df['sales'],errors='coerce')
df["store_nbr"] = df["store_nbr"].astype("int32")
df.info()
```

```python
print(df["sales"].describe())

print(
    df.nlargest(10, "sales")[
        ["date", "store_nbr", "family", "sales"]
    ]
)
```

```python
os.makedirs(f"{base_dir}/data/processed", exist_ok=True)
df.to_csv(f"{base_dir}/data/processed/train_processed2.csv", index=False)
```
